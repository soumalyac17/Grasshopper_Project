```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# Number of random points to generate per frame
n_points_per_frame = 500

# Total number of frames
n_frames = 1000
```

```python
x = np.random.rand(n_points_per_frame)
y = np.random.rand(n_points_per_frame)
print(x)
print(y)
```

```python
plt.plot(x,y)
plt.plot(x, y, marker='o', linestyle = '')
```

```python
circle = x**2 + y**2 <= 1
square = x**2 + y**2 > 1

if (len(circle) + len(square)) > 0:
  pi_estimate = 4 * len(circle) / (len(circle) + len(square))
else:
  pi_estimate = 0

print(pi_estimate)
```

```python
fig, ax = plt.subplots(figsize=(6, 6))
circle_graph = plt.Circle((0, 0), 1, color='r', fill=False)
square_graph = plt.Rectangle((0, -1), 2, 2, fill=False)
ax.add_artist(circle_graph)
ax.add_artist(square_graph)
ax.set_xlim([-1.1, 1.1])
ax.set_ylim([-1.1, 1.1])
ax.set_aspect('equal', adjustable='box')

# Update the plot
ax.clear()
ax.add_artist(circle_graph)
ax.add_artist(square_graph)
ax.set_xlim([-1.1, 1.1])
ax.set_ylim([-1.1, 1.1])
ax.set_aspect('equal', adjustable='box')
ax.plot(int(circle), 'red', label='Inside')
ax.plot(int(square), 'blue', label='Outside')
ax.set_title(f'Monte Carlo Simulation of π\nPoints: {len(circle) + len(square)}, π ≈ {pi_estimate:.4f}')
ax.legend()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Number of random points to generate per frame
n_points = 3000

# Empty lists for inside circle or outside
inside_x = []
inside_y = []
outside_x = []
outside_y = []

# Generate random points
L = int(input("Enter radius of circle: "))
x = np.random.rand(n_points)
y = np.random.rand(n_points)

# Check if points are inside the circle
inside = x**2 + y**2 <= L**2
outside = x**2 + y**2 > L**2
circle = len(inside)
square = len(inside) + len(outside)

# Append points to respective lists
inside_x.extend(x[inside])
inside_y.extend(y[inside])
outside_x.extend(x[outside])
outside_y.extend(y[outside])

# Calculate the estimate of pi
if square > 0:
  pi_estimate = 4 * circle/ square
else:
  pi_estimate = 0

# Plot graph
fig, ax = plt.subplots(figsize=(6, 6))
circle_graph = plt.Circle((0, 0), L, color='black', fill=False)
square_graph = plt.Rectangle((0, L), 2, 2, fill=False)
ax.add_artist(circle_graph)
ax.add_artist(square_graph)
ax.set_xlim([0, L])
ax.set_ylim([0, L])
ax.plot(inside_x, inside_y, 'r.', label='Inside')
ax.plot(outside_x, outside_y, 'b.', label='Outside')
ax.set_title(f'Monte Carlo Simulation of π\nPoints: {len(inside_x) + len(outside_x)}, π ≈ {pi_estimate}')
ax.legend(loc='upper right')
```
